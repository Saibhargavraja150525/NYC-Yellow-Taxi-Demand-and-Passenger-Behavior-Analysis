```python
# Phase B — Data Cleaning & Feature Engineering
```

```python
# 34.	Handle missing values where appropriate.
# Handle missing values in important columns

clean_df = df.dropna(
    subset=[
        "tpep_pickup_datetime",
        "tpep_dropoff_datetime",
        "trip_distance",
        "passenger_count"
    ]
).copy()

print("Original rows:", len(df))
print("Rows after handling missing values:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
```

```python
#35.	Handle duplicate records if duplicates are identified.
# Check for duplicate records
print("Duplicate records:", clean_df.duplicated().sum())

# Remove duplicate records
clean_df = clean_df.drop_duplicates().copy()

print("Shape after removing duplicates:", clean_df.shape)
```

```python
#36.	Convert:
# tpep_pickup_datetime
# tpep_dropoff_datetime

# Convert datetime columns
clean_df["tpep_pickup_datetime"] = pd.to_datetime(
    clean_df["tpep_pickup_datetime"]
)

clean_df["tpep_dropoff_datetime"] = pd.to_datetime(
    clean_df["tpep_dropoff_datetime"]
)

# Check the data types
print(clean_df[
    ["tpep_pickup_datetime", "tpep_dropoff_datetime"]
].dtypes)
```

```python
#37.	Investigate invalid passenger counts.
# Check passenger count values
print("Passenger Count Values:")
print(clean_df["passenger_count"].value_counts().sort_index())
```

```python
#38.	Investigate zero or negative trip distances.

zero_distance = (clean_df["trip_distance"] == 0).sum()
negative_distance = (clean_df["trip_distance"] < 0).sum()

print("Zero trip distances:", zero_distance)
print("Negative trip distances:", negative_distance)
```

```python
#39.	Investigate trips where drop-off occurs before pickup.
clean_df = clean_df[clean_df["tpep_dropoff_datetime"] >= clean_df["tpep_pickup_datetime"]].copy()
```

```python
# #40.	Investigate unrealistic trip durations.
clean_df = clean_df[((clean_df["tpep_dropoff_datetime"] - clean_df["tpep_pickup_datetime"]).dt.total_seconds() / 60) <= 180].copy()
```

```python
#41.	Document every major cleaning decision.
```

```python
# 8. Required Feature Engineering

#Creates trip duration, pickup date/hour, weekday, weekend flag, speed, distance category, duration category and route.
clean_df["trip_duration_minutes"] = (
clean_df["tpep_dropoff_datetime"] -clean_df["tpep_pickup_datetime"]
).dt.total_seconds()/60
```

```python
#Pick Up date and hour

clean_df["pickup_date"] = clean_df["tpep_pickup_datetime"].dt.date
clean_df["pickup_hour"] = clean_df["tpep_pickup_datetime"].dt.hour
```

```python
# Weekend flag
clean_df["day_of_week"] = clean_df["tpep_pickup_datetime"].dt.day_name()
clean_df["is_weekend"] = clean_df["tpep_pickup_datetime"].dt.dayofweek >= 5
)
```

```python
#Speed & Distance Category

clean_df["distance_category"] = pd.cut(
clean_df["trip_distance"],
bins=[-np.inf,2,10,np.inf],
labels=["Short","Medium","Long"]
)
clean_df["duration_category"] = pd.cut(
clean_df["trip_duration_minutes"],
bins=[-np.inf,10,20,40,np.inf],
labels=["Very Short","Short","Medium","Long"]
)
clean_df["route"] = (
clean_df["PULocationID"].astype("Int64").astype(str)
+ " → " +
clean_df["DOLocationID"].astype("Int64").astype(str)
)
```
